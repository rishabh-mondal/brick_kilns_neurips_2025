```python
import os

# Folders to compare
folder1 = "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/sentinel/Bangladesh/images"
folder2 = "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/processed_data/sentinel/stratified_split/negetive_samples"

# Get PNG filenames (without path)
pngs1 = {f for f in os.listdir(folder1) if f.lower().endswith(".png")}
pngs2 = {f for f in os.listdir(folder2) if f.lower().endswith(".png")}

# Find common files
common_pngs = sorted(pngs1 & pngs2)

print(f"Found {len(common_pngs)} common .png files:\n")

for fname in common_pngs:
    path1 = os.path.join(folder1, fname)
    path2 = os.path.join(folder2, fname)
    print(f"Filename: {fname}")
    print(f"  Path in folder1: {path1}")
    print(f"  Path in folder2: {path2}")
    print("-" * 80)
```
